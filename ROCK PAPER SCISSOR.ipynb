```python
import random

def rock_paper_scissor():
    choices = ['rock', 'paper', 'scissor']

    while True:
        comp_choice = random.choice(choices)
        player_choice = input("Enter rock, paper or scissor: ").lower()

        if player_choice not in choices:
            print("Invalid input")
            continue

        print("Computer chose:", comp_choice)

        if (
            (comp_choice == "rock" and player_choice == "scissor") or
            (comp_choice == "paper" and player_choice == "rock") or
            (comp_choice == "scissor" and player_choice == "paper")
        ):
            print("Computer wins")

        elif comp_choice == player_choice:
            print("Tie")

        else:
            print("Player wins")
            break

rock_paper_scissor()
```
